```python
import os
from dotenv import load_dotenv

from autogen import AssistantAgent, UserProxyAgent

load_dotenv()

KEY = os.getenv("GROQ_API_KEY")

llm_config = {
    "config_list": [
        {
            "model": "openai/gpt-oss-120b",
            "api_key": KEY,
            "base_url": "https://api.groq.com/openai/v1",
        }
    ]
}
# Assistant Agent

assistant = AssistantAgent(
    name="assistant",

    llm_config=llm_config,

    system_message=(
        "You are a Python developer. "
        "Write clean and simple Python code. "
        "Explain your solution briefly."
    )
)

# User Proxy Agent

user_proxy = UserProxyAgent(
    name="user_proxy",

    human_input_mode="NEVER",

    code_execution_config={
        "use_docker": False
    },

    max_consecutive_auto_reply=5
)

# Start the conversation
user_proxy.initiate_chat(
    assistant,

    message=(
        "Write a Python function to check whether "
        "a number is prime. Then test the function "
        "with 7, 10 and 13."
    )
)
```

```python
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()

client = OpenAI(
    api_key=os.getenv("GROQ_API_KEY"),
    base_url="https://api.groq.com/openai/v1"
)

response = client.chat.completions.create(
    model="openai/gpt-oss-120b",
    messages=[
        {
            "role": "user",
            "content": "Say hello in one sentence."
        }
    ]
)

print(response.choices[0].message.content)
```

```python
from botbuilder.core import ActivityHandler,
    TurnContext

class HotelBot(ActivityHandler):
    async def on_message_activity(
        self, turn_context: TurnContext
    ):
        user_text = turn_context.activity.text

        # Reuse the SAME run_agent() from Unit 1 --
        # the channel changes, the agent logic doesn't
        reply = await run_agent_async(user_text)

        await turn_context.send_activity(reply)

    async def on_members_added_activity(
        self, members_added, turn_context
    ):
        for member in members_added:
            if member.id != turn_context.activity\
                    .recipient.id:
                await turn_context.send_activity(
                    "Hi! Ask me about hotel "
                    "availability or rates."
                )
```
